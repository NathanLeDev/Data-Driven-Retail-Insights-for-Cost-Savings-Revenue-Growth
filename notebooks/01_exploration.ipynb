```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df_aisles = pd.read_csv('../data/raw/aisles.csv')
df_departments = pd.read_csv('../data/raw/departments.csv')
df_orders = pd.read_csv('../data/raw/orders.csv')
df_products = pd.read_csv('../data/raw/products.csv')
df_order_products__prior = pd.read_csv('../data/raw/order_products__prior.csv')
df_order_products__train = pd.read_csv('../data/raw/order_products__train.csv')
```

```python
df_aisles.info()
```

```python
df_departments.info()
```

```python
df_orders.info()
```

```python
df_products.info()
```

```python
df_prod = df_products.merge(df_aisles, on="aisle_id", how="left") \
                    .merge(df_departments, on="department_id", how="left")
```

```python
df_prod.info()
```

```python
orders_prior_full = df_order_products__prior.merge(df_orders, on="order_id", how="left") \
                               .merge(df_prod, on="product_id", how="left")

orders_train_full = df_order_products__train.merge(df_orders, on="order_id", how="left") \
                               .merge(df_prod, on="product_id", how="left")
```

```python
df = pd.concat([orders_prior_full, orders_train_full], axis=0, ignore_index=True)
```

```python
df.to_csv("../data/processed/df_full.csv", index=False)
```
